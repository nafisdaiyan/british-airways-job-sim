```python
df = pd.read_excel('/kaggle/input/datasets/nafda1801/british-airways-internship/British Airways Summer Schedule Dataset - Forage Data Science Task 1.xlsx')
```

```python
df['seconds_since_midnight'] = df['FLIGHT_TIME'].apply(
    lambda t: (t.hour * 3600) + (t.minute * 60) + t.second if t is not None else None
)
```

```python
df['Total_Passengers']=df['FIRST_CLASS_SEATS']+ df['BUSINESS_CLASS_SEATS'] + df['ECONOMY_SEATS']
```

```python
df.head()
```

```python
df['tier1_pct']=df['TIER1_ELIGIBLE_PAX']*100/df['Total_Passengers']
df['tier2_pct']=df['TIER2_ELIGIBLE_PAX']*100/df['Total_Passengers']
df['tier3_pct']=df['TIER3_ELIGIBLE_PAX']*100/df['Total_Passengers']
df.head()
```

```python
df['Business_ratio']=df['BUSINESS_CLASS_SEATS']/df['Total_Passengers']
df['First_ratio'] = df['FIRST_CLASS_SEATS']/df['Total_Passengers']
df['Economy_ratio'] = df['ECONOMY_SEATS']/df['Total_Passengers']
```

```python
df.head()
```

```python
lookup = (
    df.groupby(
        ["TIME_OF_DAY", "HAUL", "ARRIVAL_REGION"]
    )
    .agg(
        flights=("FLIGHT_NO", "count"),
        tier1_pct=("tier1_pct", "mean"),
        tier2_pct=("tier2_pct", "mean"),
        tier3_pct=("tier3_pct", "mean")
    )
    .reset_index()
)

lookup
```

```python
for col in ["TIME_OF_DAY", "HAUL", "ARRIVAL_REGION"]:
    print(f"\n===== {col} =====")
    print(
        df.groupby(col)[
            ["tier1_pct", "tier2_pct", "tier3_pct"]
        ].mean()
    )
```

```python
summary = (
    df.groupby("ARRIVAL_REGION")[["tier3_pct", "tier2_pct", "tier1_pct"]]
    .mean()
)

summary["tier3_range"] = (
    summary["tier3_pct"].max() -
    summary["tier3_pct"].min()
)
summary["tier2_range"] = (
    summary["tier2_pct"].max() -
    summary["tier2_pct"].min()
)
summary["tier1_range"] = (
    summary["tier1_pct"].max() -
    summary["tier1_pct"].min()
)

summary
```

```python
df.groupby(
    ["HAUL", "TIME_OF_DAY"]
)[["tier1_pct", "tier2_pct", "tier3_pct"]].mean().round(3)
```

```python
weighted_lookup = (
    df.groupby("HAUL")
      .agg(
          total_pax=("Total_Passengers", "sum"),
          tier1_pax=("TIER1_ELIGIBLE_PAX", "sum"),
          tier2_pax=("TIER2_ELIGIBLE_PAX", "sum"),
          tier3_pax=("TIER3_ELIGIBLE_PAX", "sum")
      )
      .reset_index()
)

weighted_lookup["tier1_pct"] = (
    weighted_lookup["tier1_pax"] /
    weighted_lookup["total_pax"] * 100
)

weighted_lookup["tier2_pct"] = (
    weighted_lookup["tier2_pax"] /
    weighted_lookup["total_pax"] * 100
)

weighted_lookup["tier3_pct"] = (
    weighted_lookup["tier3_pax"] /
    weighted_lookup["total_pax"] * 100
)

weighted_lookup
```

```python
weighted_time = (
    df.groupby(["HAUL", "TIME_OF_DAY"])
      .agg(
          total_pax=("Total_Passengers", "sum"),
          tier1_pax=("TIER1_ELIGIBLE_PAX", "sum"),
          tier2_pax=("TIER2_ELIGIBLE_PAX", "sum"),
          tier3_pax=("TIER3_ELIGIBLE_PAX", "sum")
      )
      .reset_index()
)

for tier in [1, 2, 3]:
    weighted_time[f"tier{tier}_pct"] = (
        weighted_time[f"tier{tier}_pax"] /
        weighted_time["total_pax"] * 100
    )

weighted_time
```
